### Problem 1: Vehicle Rental System

Create two classes, `Vehicle` and `Car`, with the following requirements:

1. **`Vehicle` Class**:
   - An `__init__` method that accepts `vehicle_id`, `brand`, and `daily_rate`, and initializes them as instance variables.
   - A **class method** `rental_policy()` that prints `"All vehicles must be returned by 5 PM."`.
   - A **method** `get_rental_cost(days)` that calculates and returns the total rental cost (daily_rate * days).
   - A **method** `display()` that prints `"Vehicle ID: {vehicle_id}, Brand: {brand}, Daily Rate: ${daily_rate}"`.

2. **`Car` Subclass** (inherits from `Vehicle`):
   - An `__init__` method that accepts `vehicle_id`, `brand`, `daily_rate`, and `num_seats`:
     - Uses `super()` to initialize `vehicle_id`, `brand`, and `daily_rate` from the parent class.
     - Adds a new instance variable `num_seats`.
   - Override the `display()` method to print:
     `"Car ID: {vehicle_id}, Brand: {brand}, Seats: {num_seats}, Daily Rate: ${daily_rate}"`.
   - Override the `get_rental_cost(days)` method to apply a 10% discount for rentals of 7 days or more.
   - Implement the `__call__` method with a parameter `customer_name` to print:
     `"Car {vehicle_id} rented to {customer_name}"`.

### Problem 2: Library Management System

Create a class system for managing books in a library:

1. **Base Class `Book`**
   - **Class Variable**: `total_books` (track all books created)
   - **Instance Variables**: `title`, `author`, `isbn`
   - **Methods**:
     - `__init__`: Initialize title, author, isbn, and increment `total_books`
     - `get_info()`: Return "'{title}' by {author} (ISBN: {isbn})"
     - `__call__`: Return "Book '{title}' is now checked out!"
     - `@classmethod get_total_books()`: Return total number of books

2. **Subclass `EBook`**
   - Inherits from `Book`
   - **New Instance Variable**: `file_size` (in MB)
   - **Override** `get_info()` to include file size: "'{title}' by {author} (ISBN: {isbn}) - {file_size}MB"
   - **New Method** `download()`: Print "Downloading '{title}'... Done!"

---

**Test your implementation with:**
