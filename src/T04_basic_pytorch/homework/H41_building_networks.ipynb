## Comparison and Analysis

**Task:** Compare the models you've built and analyze their properties.

```python
# Count total parameters in each model
def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

print("Parameter Count Comparison:")
print(f"Sequential model: {count_parameters(model_sequential):,} parameters")
print(f"Class-based model: {count_parameters(model_class):,} parameters")
print(f"Multi-branch model: {count_parameters(model_branch):,} parameters")
print(f"Flexible model [128,64,32]: {count_parameters(model_flex):,} parameters")
print(f"Flexible model [32,16]: {count_parameters(model_flex_small):,} parameters")
print(f"Flexible model [256,128,64,32,16]: {count_parameters(model_flex_large):,} parameters")
```

```python
# Test all models with the same input batch
test_batch = torch.randn(32, num_features)

with torch.no_grad():  # Disable gradient computation for inference
    out1 = model_sequential(test_batch)
    out2 = model_class(test_batch)
    out3 = model_branch(test_batch)
    out4 = model_flex(test_batch)

print("\nOutput Verification:")
print(f"All outputs have correct shape: {all(o.shape == (32, num_outputs) for o in [out1, out2, out3, out4])}")
print(f"Output dtypes match: {all(o.dtype == torch.float32 for o in [out1, out2, out3, out4])}")
```
